Part A — BASIC STATISTICS

```python
# TODO 1: Write a function mean(numbers)
# Steps:
#   - Take a list of numbers
#   - Compute the average
#   - Return the result
def mean(numbers):
    # your code here
    mean = sum(numbers)/len(numbers)
    return mean
```

```python
# TODO 2: Write a function variance(numbers)
# Steps:
#   - Use mean()
#   - Compute average squared distance from the mean
def variance(numbers):
    # your code here
    mean = sum(numbers)/len(numbers)
    variance_sum = 0
    for n in numbers:
      variance_sum += (n-mean)**2
    variance = variance_sum/len(numbers)
    return variance
```

```python
# TODO 3: Write a function std_dev(numbers)
# Steps:
#   - Use math.sqrt() on variance
def std_dev(numbers):
    # your code here
    mean = sum(numbers)/len(numbers)
    variance_sum = 0
    for n in numbers:
      variance_sum += (n - mean)**2
    variance = variance_sum/len(numbers)
    standard_deviation = math.sqrt(variance)
    return standard_deviation
```

```python
# quick test
data = [2, 4, 6, 8, 10]
print("Mean:", mean(data))
print("Variance:", variance(data))
print("Std Dev:", std_dev(data))
```

Part B — CORRELATION & COVARIANCE

```python
# TODO 4: Compute covariance between two lists x and y
def covariance(x, y):
    # your code here
    covariance = sum((x[i] - mean(x)) * (y[i] - mean(y)) for i in range(len(x))) / len(x)-1
    return covariance
```

```python
# TODO 5: Compute Pearson correlation manually
def correlation(x, y):
    # your code here
    cov = covariance(x,y)
    std_x = std_dev(x)
    std_y = std_dev(y)
    cor = cov / (std_x*std_y)
    return cor
```

```python
# quick test
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
print("Covariance:", covariance(x, y))
print("Correlation:", correlation(x, y))
```

Part D — MINI CONCEPT: GRADIENT DESCENT (1D)

 Goal: Understand how gradient descent updates weights to minimize error

```python
# TODO 7: Implement a simple gradient descent for f(w) = (w - 3)^2
# Steps:
#   - Initialize w = 0
#   - Loop N times
#   - Compute gradient = 2*(w - 3)
#   - Update w = w - lr * gradient
#   - Print each iteration
def gradient_descent(learning_rate, iterations):
    # your code here
    w= 0
    for i in range(iterations):
      gradient = 2*(w-3)
      w = w - learning_rate*gradient
      print(f"Iteration {i+1}:w={w}")

gradient_descent(0.1,10)
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("Gradient Descent Visualizer")

# -------------------------------------------------------------
# Part A — Define the Function
# -------------------------------------------------------------
# TODO 1:
#   - Define a simple quadratic function f(x) = (x - 3)^2
def f(x):
  return (x-3)**2
#   - Compute its gradient analytically: f'(x) = 2(x - 3)
def gradient(x):
  return 2*(x-3)
#   - Visualize the curve using Matplotlib
x = np.linspace(-1, 7, 100)
y = (x - 3)**2
plt.plot(x, y, label="f(x) = (x-3)^2")
plt.title("Function Curve")
plt.legend()
plt.show()


# -------------------------------------------------------------
# Part B — Implement Gradient Descent
# -------------------------------------------------------------
# TODO 2:
#   - Initialize x randomly (e.g., x = np.random.randn())
#   - Set learning_rate (e.g., 0.1)
#   - Loop for N iterations
#   - Update rule: x_new = x_old - learning_rate * gradient
#   - Store each x for visualization

# your code here
x = np.random.randn()
learning_rate = 0.1
iterations = 20

x_values = [x]
y_values = [f(x)]
for i in range(iterations):
  grad = gradient(x)
  x_new = x - learning_rate*grad
  x_values.append(x_new)
  y_values.append(f(x_new))
  x = x_new
# Example checks
print("Final value of x:", x)
print("Minimum value of f(x):", f(x))


# -------------------------------------------------------------
# Part C — Visualize the Descent Path
# -------------------------------------------------------------
# TODO 3:
#   - Plot the curve f(x)
#   - Overlay the path of x values (as red dots)
#   - Show how the value converges to the minimum at x = 3

# your code here
x_curve = np.linspace(-1,7,400)
y_curve = f(x_curve)

plt.plot(x_curve,y_curve)
plt.title("f(x) = (x - 3)^2")

plt.plot(x_values, y_values, 'ro--', label="Descent Path")
plt.legend()
plt.show()


# -------------------------------------------------------------
# Part D — Experiment with Learning Rates
# -------------------------------------------------------------
# TODO 4:
#   - Try learning rates: 0.01, 0.1, 0.5
#   - Compare convergence speed & stability
#   - Plot all three paths together for comparison

# your code here
learning_rates = [0.01,0.1,0.5]
iterations = 20
np.random.seed(24)
x_initial = np.random.randn()

x_curve= np.linspace(-1,7,400)
y_curve = f(x_curve)
plt.plot(x_curve, y_curve, label="f(x) = (x-3)^2")
for l in learning_rates:
  x = x_initial
  x_values = [x]
  y_values = [f(x)]

  for i in range(iterations):
    grad = gradient(x)
    x_new = x - l*grad
    x= x_new
    x_values.append(x)
    y_values.append(f(x))
  plt.plot(x_values,y_values)
  plt.plot(x_values, y_values, 'o--', label=f"LR={l}")

plt.title("Gradient Descent with Different Learning Rates")
plt.legend()
plt.show()
```
