```python
# TODO: sum of 1..n (use a loop)
def sum_1_to_n(n: int) -> int:
    # your code here
    total = 0
    i = 1
    while i < n+1:
      total +=i
      i+=1
    return total

# TODO: factorial n! (1*2*...*n), assume n >= 0; factorial(0) = 1
def factorial(n: int) -> int:
    # your code here
    total = 1
    i = 1
    while i < n+1:
      total = i*total
      i += 1
    return total

# TODO: return list of first n Fibonacci numbers starting [0,1,1,2,...]
def fibonacci(n: int):
    # hint: start with a=0, b=1 and loop n times
    # your code here
    sequence = []
    a,b = 0,1
    for i in range(n):
      sequence.append(a)
      a, b = b, a+b
    return sequence

assert sum_1_to_n(5) == 15
assert factorial(0) == 1 and factorial(5) == 120
assert fibonacci(1) == [0]
assert fibonacci(6) == [0, 1, 1, 2, 3, 5]
print("A3 passed ✅")
```

Part D — MINI PROJECT: TEXT CLEANER & ANALYZER

```python
import string

# Step 1: Read file (create sample.txt in Colab manually or use provided one)
with open("sample.txt","r") as f:
  text = f.read().lower()
# Step 2: Clean punctuation
for clean in string.punctuation:
  text = text.replace(clean,"")

# Step 3: Split into words and lines
words = text.split()
lines = text.splitlines()

# Step 4: Compute stats
unique_words = set(words)
word_frequency = {w:words.count(w) for w in unique_words }

# Step 5: Find most frequent word
most_frequent = max(word_frequency,key = word_frequency.get)

# Step 6: Write output summary
with open("sample.txt", "w") as f:
  f.write(f"Total words are: {len(words)}\n")
  f.write(f"Total lines are: {len(lines)}\n")
  f.write(f"Word frequency is: {word_frequency}\n")
  f.write(f"Most frequent word is: {most_frequent}\n")
print("Summary written to sample.txt ✅")
```
