```python
import pandas as pd
import numpy as np

print("=== Week 3 — Advanced Pandas ===")

# -------------------------------------------------------------
# Part A — Handle Missing Values
# -------------------------------------------------------------
# TODO 1: Load Employees.csv
employees = pd.read_csv('Employees.csv')
# TODO 2: Detect missing values (isnull().sum())
employees.isnull().sum()
# TODO 3: Fill or drop NaN values appropriately
# employees.fillna() - IF there are any nulls
# employees.dropna()m - IF there are any nulls
# TODO 4: Remove duplicates if any
duplicates = employees.duplicated()
print(duplicates)

```

```python
# -------------------------------------------------------------
# Part B — Merge with Department Info
# -------------------------------------------------------------
# TODO 5: Load Departments.csv
dept = pd.read_csv('Departments.csv')

# TODO 6: Merge with Employees on Department
inner_merge = employees.merge(dept, on='Department', how="inner")
outer_merge = employees.merge(dept, on='Department', how="outer")
merged = employees.merge(dept, on='Department', how="left")
right_merge = employees.merge(dept, on='Department', how="right")
# TODO 7: Print merged shape and sample rows
print(merged.shape)
print(merged.head(5))
```

```python

# -------------------------------------------------------------
# Part C — Pivot & Melt
# -------------------------------------------------------------
# TODO 8: Create a pivot table (avg Salary by Department × WorkType)
# TODO 9: Convert it back using melt()
pivot_merged = pd.pivot_table(merged,values='Salary',index='Department',columns='RemoteWork',aggfunc='mean')
print(pivot_merged)
pivot_melt = pivot_merged.reset_index().melt(id_vars='Department',value_vars=pivot_merged.columns,var_name='RemoteWork',value_name="Average_Salary")
print(pivot_melt)

```

```python
# -------------------------------------------------------------
# Part D — Apply & Lambda
# -------------------------------------------------------------
# TODO 10: Use apply() to compute AnnualBonus (Salary × BonusPercent)
annual_bonus = merged.apply(lambda x: x['Salary']*x['BonusPercent'], axis=1 )
print(annual_bonus)
# TODO 11: Create a Tier column based on PerformanceScore
performance_tier = merged.apply(lambda score: "Top Performer" if score['PerformanceScore'] > 4 else ("Okay performance" if score['PerformanceScore']  == 3   else "Low Performance" ),axis=1)
print(performance_tier)
```

```python
# -------------------------------------------------------------
# Part E — Save Cleaned File
# -------------------------------------------------------------
# TODO 12: Save merged cleaned data as Employees_Cleaned.csv
merged.to_csv('Employees_Cleaned.csv')

# TODO 13: Print confirmation message
print('The dataset is now clean.')
cleaned = pd.read_csv('Employees_Cleaned.csv')
print(employees.info())
print(cleaned.info)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("=== Week 3 - Visualization ===")

# -------------------------------------------------------------
# Part A — Setup
# -------------------------------------------------------------
# TODO 1: Load Employees_Cleaned.csv
cleaned = pd.read_csv('Employees_Cleaned.csv')
cleaned = cleaned.drop(columns=['Unnamed: 0'])

# TODO 2: Set Seaborn theme
sns.set_theme(style = 'whitegrid')

# -------------------------------------------------------------
# Part B — Bar Plot
# -------------------------------------------------------------
# TODO 3: Create a barplot for average Salary by Department
plt.figure(figsize=(5,5))
sns.barplot(data = cleaned, x='Department', y= 'Salary', estimator=np.mean,ci = None)
plt.title("Average Salary By Department")
plt.show()


# -------------------------------------------------------------
# Part C — Histogram & Boxplot
# -------------------------------------------------------------
# TODO 4: Plot histogram of Age
plt.figure(figsize=(5,5))
sns.histplot(data = cleaned['Age'])
plt.title("Histogram of Employee ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
# TODO 5: Plot boxplot of Salary by JobTitle
plt.figure(figsize=(5,5))
sns.boxplot(data = cleaned, x='JobTitle', y= 'Salary' )
plt.title("Boxplot of Salary vs JobTitle")
plt.xlabel("Job Title")
plt.ylabel("Salary")
plt.xticks(rotation = 90)
# -------------------------------------------------------------
# Part D — Scatterplot
# -------------------------------------------------------------
# TODO 6: Create a scatterplot for ExperienceYears vs Salary
plt.figure(figsize=(5,5))
sns.regplot(data = cleaned, x='ExperienceYears', y= 'Salary',scatter_kws={'alpha': 0.6})
plt.title("Scatterplot of ExperienceYears vs Salary")
plt.xlabel("ExperienceYears")
plt.ylabel("Salary")
plt.show()

# -------------------------------------------------------------
# Part E — Correlation Heatmap
# -------------------------------------------------------------
# TODO 7: Display correlation heatmap for numeric columns
plt.figure(figsize=(10,9))
correlation = cleaned.corr(numeric_only=True)

```

```python
import pandas as pd
import numpy as np

print("=== Week 3 — Mini EDA Report ===")

# -------------------------------------------------------------
# Part A — Snapshot
# -------------------------------------------------------------
# TODO 1: Load Employees_Cleaned.csv
merged.to_csv("Employees_Cleaned.csv", index=False)
cleaned = pd.read_csv('Employees_Cleaned.csv')

# TODO 2: Print shape, columns, and head
print(cleaned.shape)
print(cleaned.columns)
print(cleaned.head(5))
# -------------------------------------------------------------
# Part B — KPIs
# -------------------------------------------------------------
# TODO 3: Compute key metrics
#   - Total employees
#   - Avg Salary
#   - Avg Experience
#   - Top Performer %
total_employees = cleaned['EmpID'].count()
print(total_employees)
avg_salary = cleaned['Salary'].mean()
print(avg_salary)
avg_experience = cleaned['ExperienceYears'].mean()
print(avg_experience)
top_performers = (cleaned['PerformanceScore']>4).mean()*100
print(round(top_performers,2),"%")
# -------------------------------------------------------------
# Part C — Grouped Insights
# -------------------------------------------------------------
# TODO 4: Department-level averages for Salary & PerformanceScore
# TODO 5: Sort departments by average salary
dept_avg = cleaned.groupby('Department')[['Salary','PerformanceScore']].mean()
print(dept_avg)
dept_sort = dept_avg.sort_values(by='Salary',ascending = False)
print(dept_sort)


# -------------------------------------------------------------
# Part D — Pivot Example
# -------------------------------------------------------------
# TODO 6: Create pivot table (Salary by Department × Gender)
salary_pivot = pd.pivot_table(cleaned,values='Salary',index='Department',columns='Gender',aggfunc='mean')
print(salary_pivot)

# -------------------------------------------------------------
# Part E — Export Summary
# -------------------------------------------------------------
# TODO 7: Export department summary CSV
dept_sort.to_csv('Summary.csv',index=True)
# TODO 8: Print confirmation message
print('Summary has been updated')
```
